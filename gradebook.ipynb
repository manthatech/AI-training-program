```python
# Grade book list
grade_book = []

# Function to add student
def add_student(name, scores):

    # Calculate average
    average = sum(scores) / len(scores)

    # Determine grade
    if average >= 90:
        grade = "A"

    elif average >= 80:
        grade = "B"

    elif average >= 70:
        grade = "C"

    elif average >= 60:
        grade = "D"

    else:
        grade = "F"

    # Create dictionary
    student = {
        "name": name,
        "scores": scores,
        "average": round(average, 2),
        "grade": grade
    }

    # Add to grade book
    grade_book.append(student)

    # Print confirmation
    print(f"Added {name} | Avg: {average:.1f} | Grade: {grade}")


def get_class_stats():
    if grade_book is None:
        print("No students yet")
    averages = []

    for student in grade_book:
        #print(student)
        averages.append(student["average"])
    print(f"\nTotal students", len(grade_book))
    print(f"Hightest average is", max(averages))
    print(f"Lowest average is", min(averages))
    print(f"class average",sum(averages)/len(averages))

def get_top_students(n=3):
    sorted_students = sorted(
        grade_book,
        key=lambda s: s["average"],
        reverse=True
    )

    print(f"\nTop {n} Students:")

    rank = 1

    for student in sorted_students[:n]:

        print(
            f"{rank}. {student['name']} "
            f"- {student['average']:.2f} "
            f"({student['grade']})"
        )

        rank += 1


def grade_distribution():

    distribution = {}

    for student in grade_book:

        grade = student["grade"]

        distribution[grade] = distribution.get(grade, 0) + 1

    return distribution

# Function calls
add_student("Adhiran", [95, 90, 87])
add_student("Aravind", [90, 96, 80])
add_student("Gayatri", [78, 60, 80])
add_student("Mamatha", [88, 62, 79])
add_student("siri", [60, 35, 88])
add_student("ira", [33, 26, 28])


get_class_stats()
get_top_students(n=3)
print("\nGrade Distribution:",grade_distribution())
```
